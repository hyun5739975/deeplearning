```python
import tensorflow as tf
from sklearn.model_selection import train_test_split
```

```python
(train_input, train_target), (test_input, test_target) \
    = tf.keras.datasets.fashion_mnist.load_data()
```

```python
# 정규화
train_scaled = train_input / 255.0
test_scaled = test_input / 255.0

# 입력데이터: 2 dim -> 1 dim 변환
train_scaled = train_scaled.reshape(-1, 28 * 28)
train_scaled.shape
```

### 신경망 구성 방법 2

```python
dense1 = tf.keras.layers.Dense(units      = 50
                             , activation = 'sigmoid'
                             , name       = 'hidden1')

dense2 = tf.keras.layers.Dense(units      = 100
                             , activation = 'sigmoid'
                             , name       = 'hidden2')

dense3 = tf.keras.layers.Dense(units      = 10
                             , activation = 'sigmoid'
                             , name       = 'hidden3')

output = tf.keras.layers.Dense(units      = 10
                             , activation = 'softmax'
                             , name       = 'output')

```

```python
model = tf.keras.Sequential([tf.keras.Input(shape=(28*28,))
                          , dense1
                          , dense2
                          , dense3
                          , output])
```

```python
model.summary()
```

### 신경망 구성 방법 3

```python
model = tf.keras.Sequential([tf.keras.Input(shape=(28*28,))
                          , tf.keras.layers.Dense(
                               units      = 50
                             , activation = 'sigmoid'
                             , name       = 'hidden1')                        
                          , tf.keras.layers.Dense(
                               units      = 100
                             , activation = 'sigmoid'
                             , name       = 'hidden2')         
                          , tf.keras.layers.Dense(
                               units      = 10
                             , activation = 'sigmoid'
                             , name       = 'hidden3')     
                          , tf.keras.layers.Dense(
                               units      = 10
                             , activation = 'softmax'
                             , name       = 'output')])

model.summary()
```

```python
model.compile(loss='sparse_categorical_crossentropy'
            , optimizer='sgd'
            , metrics=['accuracy'])
```

```python
loss = model.fit(train_scaled, train_target, epochs=100)
```

```python
test_scaled = test_scaled.reshape(-1, 28 * 28)
```

```python
model.evaluate(test_scaled, test_target)
```
