```python
import tensorflow as tf
from sklearn.model_selection import train_test_split
```

```python
(train_input, train_target), (test_input, test_target) \
    = tf.keras.datasets.fashion_mnist.load_data() # 6만개 데이터
```

```python
# 정규화
train_scaled = train_input / 255.0
test_scaled  = test_input / 255.0

train_scaled = train_scaled.reshape(-1, 28 * 28)
test_scaled  = test_scaled.reshape(-1, 28 * 28)
```

```python
train_scaled, val_scaled, train_target, val_target = \
    train_test_split(train_scaled
                   , train_target
                   , test_size    = 0.2
                   , random_state = 42)
```

```python
print(train_scaled.shape, val_scaled.shape)
```

```python
model = tf.keras.Sequential()

# Xavier 정규분포 초기값 설정기
model.add(tf.keras.Input(shape = (28 * 28,)))

model.add(tf.keras.layers.Dense(
    units              = 1024
  , activation         = 'relu'
  , kernel_initializer = 'glorot_normal'))

model.add(tf.keras.layers.Dropout(0.3)) # 70% 오버피팅 개선

model.add(tf.keras.layers.Dense(
    units              = 1024
  , activation         = 'relu'
  , kernel_initializer = 'glorot_normal'))

model.add(tf.keras.layers.Dropout(0.3)) # 70%

model.add(tf.keras.layers.Dense(
    units              = 512
  , activation         = 'relu'
  , kernel_initializer = 'glorot_normal'))

model.add(tf.keras.layers.Dropout(0.3)) # 70%

model.add(tf.keras.layers.Dense(
    units              = 10
  , activation         = 'softmax'
  , kernel_initializer = 'glorot_normal'))
```

```python
model.summary()
```

```python
# Adam(learning_rate=0.001)는 Adam 옵티마이저가 가장 안정적이고 빠르게 정답(최적값)을 찾아가도록 검증된 기본값
model.compile(optimizer = tf.keras.optimizers.Adam(learning_rate=0.001)
            , loss      = 'sparse_categorical_crossentropy'
            , metrics   = ['accuracy'])
                        
model.fit(train_scaled
        , train_target
        , epochs          = 50
        , batch_size      = 1000
        , validation_data = (val_scaled, val_target))
```

```python
# evaluate -> 처음 보는 시험 문제(테스트 데이터)'를 얼마나 잘 맞히는지 최종 성적을 채점하는 함수
model.evaluate(test_scaled, test_target)
```
