```python
import easyocr

reader = easyocr.Reader(["ko", "en"])

# ▶ 이미지 파일 경로 지정 (예: C드라이브의 images 폴더 안)
image_path = r"C:\Users\user\Pictures\Screenshots\sample.png"

result = reader.readtext(image_path, detail=0)
print(result)
```

```python
from PIL import Image, ImageOps
import pytesseract

pytesseract.pytesseract.tesseract_cmd = (
    r"C:\Program Files\Tesseract-OCR\tesseract.exe"
)
image_path = r"C:\Users\user\Pictures\Screenshots\sample.png"

# # 이미지를 흑백(L)으로 변환하고 크기 2배 확대
img = Image.open(image_path).convert("L")
img = img.resize((img.width * 2, img.height * 2))

# --psm 6: 단일 텍스트 블록으로 간주하고 읽기
text = pytesseract.image_to_string(img, lang="kor+eng", config="--psm 6")
print("결과:", repr(text))
```
